## What is ANN?
An Artificial Neural Network (ANN) is a model made up of connected computational units organized into layers.

A basic ANN has three types of layers:
1. Input Layer(Recieves the input features)
2. Hidden Layer(Learn patterns using weights, biases and activation functions)
3. Output Layer(Produces the final prediction)

# What happens inside a neuron?

first it calculates: z = w1x1 + w2x2 + b

Then applies a activation function: a = f(z)

Here:
- \(x_1,x_2\): inputs
- \(w_1,w_2\): learnable weights
- \(b\): learnable bias
- \(z\): weighted sum before activation
- \(f\): activation function
- \(a\): output of the neuron
For example, using ReLU:
\[
f(z)=\max(0,z)
\]

If \(z=-3\), ReLU produces \(0\). If \(z=4\), ReLU produces \(4\).
The combination of a linear transformation and a nonlinear activation is a key building block of many neural networks.

```python
# Let's run a neuron in pytorch
import torch
import torch.nn as nn

# Two input and one output feature
layer = nn.Linear(2,1)

# one example with two features
x = torch.tensor([2.0, 3.0])

z = layer(x)

print("Input:", x)
print("Weight:", layer.weight)
print("Bias:", layer.bias)
print("Output:", z)

print("="*60)
print("nn.Linear(2,1) creates a layer that accepts two input features and produce one output")
print("The output is calculated as z = w^T.x + b")
print("Weights and biases are initialize automatically so output will vary.")
print("="*60)
print("Now let's add ReLU:......")
```

```python
relu = nn.ReLU()

a = relu(z)


print("Before activation:", z)
print("After activation:", a)
```

**Important**: nn.Linear does not automatically apply ReLU. We explicitly add the activation as a separate layer.

##Building an ANN with hidden layer

my architechture:
- input features: 2
- Linear(2,4): means 4 hidden neurons
- ReLU() NonLinear Activation
- Linear(4,1): 1 Output

```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(2,4),
    nn.ReLU(),
    nn.Linear(4,1)
)

X = torch.tensor([
    [2.0, 3.0],
    [1.0, 4.0],
    [5.0, 2.0]
])

predictions = model(X)

print("Input shape:", X.shape)
print("Output shape:", predictions.shape)
print(predictions)
```
