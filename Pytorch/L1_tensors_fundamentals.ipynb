## Now let's learn some tensor operations now which are important in calculations

```python
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(f"Addition: {a+b}")
print(f"Multiplication: {a*b}")
```

```python
# Let's do some matrix multiplication which is important we have to know some linear algebra basics here
a = torch.tensor([
    [1,2],
    [3,4]
])

b = torch.tensor([
    [5,6],
    [7,8]
])

print(a@b)
```

For matrix multiplication we can use two things:
- a@b   or
- torch.matmul(a,b)

```python
x = torch.randn(3, 4)

print("Tensor:")
print(x)

print("Dimensions:", x.ndim)
print("Shape:", x.shape)
print("Data type:", x.dtype)
print("Number of elements:", x.numel())
```

### Reshape function do ?
here we can just make dimensions according to our will e.g 2,3 means two dimensions if i write 2,4,3 then it will be three that can be also our will

```python
x = torch.arange(12)
y = x.reshape(3,4)

print(y)
print(y.shape)
```

```python
y = x.reshape(2,2,3)
print(y)
```

### Flatten do ?
Everything flattens into 1 dimensions

```python
x = torch.randn(2, 3, 4)

print(x.shape)
```

```python
y = x.flatten()
print(y.shape)
```

### unsqueeze do?
Before:

[1, 2, 3]

shape = (3,)


After unsqueeze(0):

[[1, 2, 3]]

shape = (1, 3)

```python
x = torch.tensor([1,2,3])
print(x)
print(x.shape)
```

```python
y = x.unsqueeze(0)
print(y)
print(y.shape)
```

```python
y = x.unsqueeze(1)
print(y)
print(y.shape)
```

While squeeze is completely opposite of unsqueeze it actually delete one dimension:

```python
x = torch.tensor([[1,2,2,3]])
print(x)
print(x.shape)
```

```python
y = x.squeeze(0)
print(y)
print(y.shape)
```
