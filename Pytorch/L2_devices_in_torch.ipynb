```python
import torch
import numpy as np
```

We can convert numpy to torch and torch to numpy

```python
arr = np.array([1, 2, 3, 4])

print(arr)
print(type(arr))
```

```python
x = torch.from_numpy(arr)

print(x)
print(x.dtype)
print(type(x))
```

```python
# e.g torch to numpy
x = torch.tensor([1, 2, 3, 4])

arr = x.numpy()

print(arr)
print(type(arr))
```

If we transform from torch to numpy they share same memory so it is a problem with that. Changing one might affect another.

## important
only numpy share same memory with torch but torch always creates a separate memory.

```python
arr = np.array([1, 2, 3, 4])

x = torch.from_numpy(arr)

arr[0] = 100

print(arr)
print(x)
```
